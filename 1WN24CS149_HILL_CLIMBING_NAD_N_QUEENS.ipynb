```python

def calculate_cost(state):
    """
    Calculate the number of pairs of queens attacking each other.
    One queen is placed in each column.
    state[i] represents the row of the queen in column i.
    """
    n = len(state)
    conflicts = 0

    for i in range(n):
        for j in range(i + 1, n):
            # Same row or same diagonal
            if state[i] == state[j] or abs(state[i] - state[j]) == abs(i - j):
                conflicts += 1

    return conflicts


def get_neighbors(state):
    """
    Generate neighbors by swapping the row positions
    of every pair of queens.
    """
    n = len(state)
    neighbors = []

    for i in range(n):
        for j in range(i + 1, n):
            neighbor = state.copy()
            neighbor[i], neighbor[j] = neighbor[j], neighbor[i]

            neighbors.append((i, j, neighbor, calculate_cost(neighbor)))

    return neighbors


def hill_climbing(initial_state):
    current = initial_state.copy()
    step = 1

    print("Initial State:", current)
    print("Initial Cost:", calculate_cost(current))

    while True:
        current_cost = calculate_cost(current)

        print("\n" + "=" * 50)
        print(f"STEP {step}")
        print("Current State:", current)
        print("Current Cost:", current_cost)

        neighbors = get_neighbors(current)

        print("\nNeighbor states and costs:")

        # Neighbors are generated in ascending column order.
        # min() keeps the first neighbor when costs are tied.
        for i, j, state, cost in neighbors:
            print(
                f"Swap (x{i}, x{j}) -> {state}, Cost = {cost}"
            )

        best_neighbor = min(neighbors, key=lambda item: item[3])
        i, j, next_state, next_cost = best_neighbor

        print("\nBest Neighbor:", next_state)
        print("Best Neighbor Cost:", next_cost)

        # Stop if no improvement is possible.
        if next_cost >= current_cost:
            print("\nAlgorithm terminated.")
            print("Final State:", current)
            print("Final Cost:", current_cost)

            if current_cost == 0:
                print("Goal state reached!")
            else:
                print("Local minimum reached; goal not found.")

            return current

        print(f"Choose neighbor by swapping x{i} and x{j}.")
        current = next_state
        step += 1


# Main program
initial_state = [3, 1, 2, 0]
solution = hill_climbing(initial_state)
```
