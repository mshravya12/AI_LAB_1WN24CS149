```python

# Forward Chaining in First Order Logic

# Initial Knowledge Base (Facts)
facts = {
    ("Owns", "A", "T1"),
    ("Missile", "T1"),
    ("American", "Robert"),
    ("Enemy", "A", "America")
}

# Rules: (Premises, Conclusion)
rules = [
    (
        [("Missile", "x")],
        ("Weapon", "x")
    ),
    (
        [("Missile", "x"), ("Owns", "A", "x")],
        ("Sells", "Robert", "x", "A")
    ),
    (
        [("Enemy", "x", "America")],
        ("Hostile", "x")
    ),
    (
        [
            ("American", "p"),
            ("Weapon", "q"),
            ("Sells", "p", "q", "r"),
            ("Hostile", "r")
        ],
        ("Criminal", "p")
    )
]


# Check whether a symbol is a variable
def is_variable(x):
    return x.islower()


# Match a rule premise with a fact
def unify(pattern, fact, bindings):
    if pattern[0] != fact[0]:
        return None

    if len(pattern) != len(fact):
        return None

    new_bindings = bindings.copy()

    for p, f in zip(pattern[1:], fact[1:]):
        if is_variable(p):
            if p in new_bindings:
                if new_bindings[p] != f:
                    return None
            else:
                new_bindings[p] = f
        elif p != f:
            return None

    return new_bindings


# Find all possible variable bindings for premises
def match_premises(premises, facts, bindings=None):
    if bindings is None:
        bindings = [{}]

    for premise in premises:
        new_bindings = []

        for binding in bindings:
            for fact in facts:
                result = unify(premise, fact, binding)

                if result is not None:
                    new_bindings.append(result)

        bindings = new_bindings

    return bindings


# Replace variables with their values
def substitute(conclusion, bindings):
    return tuple(
        bindings.get(x, x) if is_variable(x) else x
        for x in conclusion
    )


# Forward Chaining Algorithm
def forward_chaining(facts, rules):
    iteration = 1

    while True:
        new_facts = set()

        for premises, conclusion in rules:
            bindings_list = match_premises(premises, facts)

            for bindings in bindings_list:
                new_fact = substitute(conclusion, bindings)

                if new_fact not in facts:
                    new_facts.add(new_fact)

        if not new_facts:
            break

        print("\nIteration", iteration)

        for fact in new_facts:
            print("Inferred:", fact)

        facts.update(new_facts)
        iteration += 1

    return facts


# Execute Forward Chaining
result = forward_chaining(facts, rules)

# Check the goal
goal = ("Criminal", "Robert")

print("\nFinal Result:")

if goal in result:
    print("Robert is Criminal: TRUE")
else:
    print("Robert is Criminal: FALSE")
print("shravya")

```
