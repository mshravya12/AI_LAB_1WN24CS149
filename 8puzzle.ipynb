```python
_ = "_"

# Sample starting puzzle
start = (
    ("1", "4", "2"),
    ("3", "_", "5"),
    ("6", "7", "8")
)

# Goal puzzle
goal = (
    ("_", "1", "2"),
    ("3", "4", "5"),
    ("6", "7", "8")
)


# Depth Limited DFS
def dls(state, depth, path):

    if state == goal:
        return path

    if depth == 0:
        return None

    # Find blank
    for r in range(3):
        for c in range(3):
            if state[r][c] == "_":
                break
        else:
            continue
        break

    moves = [
        (-1, 0, "Up"),
        (1, 0, "Down"),
        (0, -1, "Left"),
        (0, 1, "Right")
    ]

    for dr, dc, move in moves:

        nr = r + dr
        nc = c + dc

        if 0 <= nr < 3 and 0 <= nc < 3:

            new = [list(row) for row in state]

            # Swap blank
            new[r][c], new[nr][nc] = \
                new[nr][nc], new[r][c]

            new = tuple(tuple(row) for row in new)

            # Avoid repeated states
            if new not in [x[0] for x in path]:

                answer = dls(
                    new,
                    depth - 1,
                    path + [(new, move)]
                )

                if answer:
                    return answer

    return None


# Iterative Deepening DFS
def iddfs(start):

    for depth in range(1, 10):

        answer = dls(
            start,
            depth,
            [(start, "Start")]
        )

        if answer:
            return answer

    return None


# Solve
solution = iddfs(start)


# Print solution
if solution:

    print("Solution found!")
    print("Total steps:", len(solution) - 1)
    print()

    for i, (state, move) in enumerate(solution):

        print("Step:", i)
        print("Depth:", i)
        print("Move:", move)

        for row in state:
            print(" ".join(row))

        print()

else:
    print("No solution")
print("1WN24CS149")


```

```python
import numpy as np

_ = "_"

start = np.array([
    ["5", "4", _],
    ["6", "1", "8"],
    ["7", "3", "2"]
])

goal = np.array([
    [_, "1", "2"],
    ["3", "4", "5"],
    ["6", "7", "8"]
])


def dfs(start, goal):

    stack = [start]
    visited = set()

    while stack:

        state = stack.pop()

        key = tuple(state.flatten())

        if key in visited:
            continue

        visited.add(key)

        print(state)
        print()

        if np.array_equal(state, goal):
            print("Goal reached!")
            return

        r, c = np.argwhere(state == "_")[0]

        # Possible moves
        for dr, dc in [(-1,0), (1,0), (0,-1), (0,1)]:

            nr = r + dr
            nc = c + dc

            if 0 <= nr < 3 and 0 <= nc < 3:

                new = state.copy()

                new[r,c], new[nr,nc] = new[nr,nc], new[r,c]

                stack.append(new)

    print("No solution")


dfs(start, goal)
print("1WN24CS149")
```
