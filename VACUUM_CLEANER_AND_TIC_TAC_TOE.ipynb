```python
# Tic-Tac-Toe Game

board = [" " for _ in range(9)]

def display_board():
    print()
    print(board[0] + " | " + board[1] + " | " + board[2])
    print("--+---+--")
    print(board[3] + " | " + board[4] + " | " + board[5])
    print("--+---+--")
    print(board[6] + " | " + board[7] + " | " + board[8])
    print()


def check_winner(player):
    winning_combinations = [
        (0, 1, 2),
        (3, 4, 5),
        (6, 7, 8),
        (0, 3, 6),
        (1, 4, 7),
        (2, 5, 8),
        (0, 4, 8),
        (2, 4, 6)
    ]

    for a, b, c in winning_combinations:
        if board[a] == board[b] == board[c] == player:
            return True

    return False


def check_draw():
    return " " not in board


# Main game
current_player = "X"

while True:

    display_board()

    print("Player", current_player, "turn")

    try:
        position = int(input("Enter position (1-9): ")) - 1
    except ValueError:
        print("Please enter a number from 1 to 9.")
        continue

    if position < 0 or position > 8:
        print("Invalid position!")
        continue

    if board[position] != " ":
        print("That position is already taken!")
        continue

    board[position] = current_player

    # Check winner
    if check_winner(current_player):
        display_board()
        print("Player", current_player, "wins!")
        break

    # Check draw
    if check_draw():
        display_board()
        print("It's a draw!")
        break

    # Change player
    if current_player == "X":
        current_player = "O"
    else:
        current_player = "X"
```

```python
# Simple Reflex Vacuum Cleaner Agent

# -------- INITIAL STATE --------

print("----- INITIAL STATE -----")

location_A = input("Enter status of Location A (CLEAN/DIRTY): ").upper()
location_B = input("Enter status of Location B (CLEAN/DIRTY): ").upper()
current_position = input("Enter current position of robot (A/B): ").upper()

print("\nInitial State:")
print("Location A :", location_A)
print("Location B :", location_B)
print("Robot Position :", current_position)


# -------- VACUUM CLEANER AGENT --------

while location_A == "DIRTY" or location_B == "DIRTY":

    if current_position == "A":

        if location_A == "DIRTY":
            print("\nRobot at A → SUCK")
            location_A = "CLEAN"

        else:
            print("\nLocation A is CLEAN → MOVE RIGHT")
            current_position = "B"

    elif current_position == "B":

        if location_B == "DIRTY":
            print("\nRobot at B → SUCK")
            location_B = "CLEAN"

        else:
            print("\nLocation B is CLEAN → MOVE LEFT")
            current_position = "A"


# -------- FINAL STATE --------

print("\n----- FINAL STATE -----")

print("Location A :", location_A)
print("Location B :", location_B)
print("Robot Position :", current_position)

print("\nCleaning completed!")
```
