```python
# First Order Logic - Unification Algorithm

# Check whether a term is a variable
def is_variable(x):
    return isinstance(x, str) and x.islower()


# Apply existing substitutions
def apply_substitution(x, subst):
    if is_variable(x) and x in subst:
        return apply_substitution(subst[x], subst)

    if isinstance(x, tuple):
        return (x[0], tuple(apply_substitution(a, subst)
                            for a in x[1]))

    return x


# Check whether a variable occurs in a term
def occurs(var, term, subst):
    term = apply_substitution(term, subst)

    if var == term:
        return True

    if isinstance(term, tuple):
        return any(occurs(var, a, subst)
                   for a in term[1])

    return False


# Unification of two terms
def unify_terms(x, y, subst):

    x = apply_substitution(x, subst)
    y = apply_substitution(y, subst)

    # If both terms are identical
    if x == y:
        return True

    # If first term is a variable
    if is_variable(x):
        if occurs(x, y, subst):
            return False
        subst[x] = y
        return True

    # If second term is a variable
    if is_variable(y):
        if occurs(y, x, subst):
            return False
        subst[y] = x
        return True

    # Check function symbols and arguments
    if isinstance(x, tuple) and isinstance(y, tuple):

        if x[0] != y[0] or len(x[1]) != len(y[1]):
            return False

        for a, b in zip(x[1], y[1]):
            if not unify_terms(a, b, subst):
                return False

        return True

    return False


# Unification of two predicates
def unify(p1, p2):

    subst = {}

    # Check predicate symbols
    if p1[0] != p2[0]:
        return None

    # Check number of arguments
    if len(p1[1]) != len(p2[1]):
        return None

    # Unify corresponding arguments
    for a, b in zip(p1[1], p2[1]):

        if not unify_terms(a, b, subst):
            return None

    return {v: apply_substitution(t, subst)
            for v, t in subst.items()}


# Main Program
p1 = ("Eats", ("x", "Apple"))
p2 = ("Eats", ("Riya", "y"))

result = unify(p1, p2)

if result is not None:
    print("Unification Successful")
    print("Substitution Set:", result)
else:
    print("Unification Failed")
```
